# LeetCode 刷题篇

### ✅算法新手如何准备？
如果你是非科班转行，算法基础薄弱，强烈建议在刷LeetCode之前，系统学习下面资料：  

https://www.hello-algo.com/

这是一本对新手非常友好的算法入门资料，作者用可视化动画的方式讲解了典型的数据结构和算法，同时可以进行可视化step by step的调试。  

大家在学习时，可以配合[LeetCode基础算法题（必须会）](./02LeetCode基础算法题(必须会).ipynb)一起食用，效果更佳。
>以上两份资料要求的掌握程度为五颗星，必须能够手写代码，熟练掌握。
