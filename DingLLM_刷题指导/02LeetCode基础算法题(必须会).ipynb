## ✅选择排序、插入排序、冒泡排序  
>这三个的时间复杂度都是O(N^2)


```python
# 选择排序思想：每轮从未排序区间选择最小的元素，将其放到已排序区间的末尾。
# 时间复杂度：O(N^2)
# 算法特点：排序时间和初始顺序无关，即使初始已经排好序了，也要耗费O(N^2)的时间
def selection_sort(nums):
    N = len(nums)
    for i in range(N):
        # 选择未排序区间中最小的元素
        min_idx = i
        for j in range(i+1,N):
            if nums[j] < nums[min_idx]:
                min_idx = j
        # 将其放到已排序区间的末尾
        nums[i], nums[min_idx] = nums[min_idx], nums[i]
    return nums

nums = [7, 6, 8, 5, 2, 1, 3, 4, 0, 9, 10]
print(nums)
selection_sort(nums)
```

```python
# 冒泡排序（bubble sort）通过连续地比较与交换相邻元素实现排序。这个过程就像气泡从底部(nums 左边)升到顶部(nums 右边)一样，因此得名冒泡排序。
# 时间复杂度：O(N^2)
def bubble_sort(nums):
    N = len(nums)
    for i in range(N):
        # 是否发生了交换相邻元素
        isSwapped = False
        for j in range(0, N-i-1):
            if nums[j] > nums[j+1]:
                # 交换元素
                nums[j], nums[j+1] = nums[j+1], nums[j]
                isSwapped = True
        # 如果没有交换，数组已经有序，退出循环
        if not isSwapped:
            break
        
    return nums

nums = [7, 6, 8, 5, 2, 1, 3, 4, 0, 9, 10]
print(nums)
bubble_sort(nums)
```

```python
# 插入排序思想：就像打牌时整理牌，小的插到左边，大的插到右边
# 算法特点：最坏情况下时间复杂度O(N^2)
# 依赖于初始排列顺序，特别适合于部分有序的数组
# 完全有序的情况下，复杂度能达到O(N)，但逆序时效率降到O(N^2)
def insertion_sort(nums: list[int]):
    """ 插入排序"""
    # 外循环：已排序区间为 [0, i-1]
    for i in range(1, len(nums)):
        base = nums[i]
        j = i - 1
        # 内循环：将 base 插入到已排序区间 [0, i-1] 中的正确位置
        while j >= 0 and base < nums[j]:
            nums[j + 1] = nums[j]
            # 将 nums[j] 向右移动一位
            j -= 1
        nums[j + 1] = base  # 将 base 赋值到正确位置
        
nums = [7, 6, 8, 5, 2, 1, 3, 4, 0, 9, 10]
print(nums)
insertion_sort(nums)
nums
```

## ✅归并排序、快速排序  
>这两个都是基于分治的算法


```python
# 归并排序是一种基于分治法的排序算法，它将数组递归地分成两个子数组，分别排序后再合并。
# 时间复杂度为O(n log n)，因为数组每次都被二分，合并的时间复杂度为O(n)。
# 空间复杂度为O(n)，因为需要额外的空间来存储临时数组。

def findMid(nums):
    return len(nums) // 2

def mergeTwoNums(nums1, nums2):
    i,j = 0,0
    result = []
    while i < len(nums1) and j < len(nums2):
        if nums1[i] <= nums2[j]:  # 使用 <= 保证稳定性
            result.append(nums1[i])
            i+=1
        else:
            result.append(nums2[j])
            j+=1
        
    result += nums1[i:] if i < len(nums1) else nums2[j:]
        
    return result

def merge_sort(nums):
    if len(nums)<2:
        return nums.copy()
    
    mid = findMid(nums)
    left_res = merge_sort(nums[:mid])
    right_res = merge_sort(nums[mid:])
    
    return mergeTwoNums(left_res, right_res)


nums = [7, 6, 8, 5, 2, 1, 3, 4, 0, 9, 10]
# print(findMid(nums))
print(nums)
merge_sort(nums)
# nums
```

```python
# 快速排序也是一种基于分治法的排序算法。它通过选择一个“枢纽”元素，将数组分成两部分，其中一部分元素小于枢纽，另一部分大于枢纽，然后递归地对两部分排序。
# 时间复杂度为O(n log n)（平均情况），因为每次分区大约将数组分成两半。最坏情况下（如已排序数组），时间复杂度为O(n²)
# 空间复杂度为O(log n)，因为递归调用栈的深度为log n
def partition(nums, left, right):
    # 哨兵划分
    pivot_idx = left #选择最左边的元素称为枢纽
    pivot = nums[pivot_idx] 
    i , j= left, right
    # 从右往左找出第一个小于pivot的数
    # 从左往右找出第一个大于pivot的数
    while i<j:
        while i <j and nums[j] >= pivot:
            j -= 1
        while i < j and nums[i] <= pivot:
            i += 1
        if i<j:
            nums[i],nums[j] = nums[j], nums[i]
    # 全都交换完毕，此时i=j
    nums[i],nums[pivot_idx] = nums[pivot_idx],nums[i]
    
    return i

def quick_sort(nums,left,right):
    if left >= right: return
    
    pivot = partition(nums, left, right)  # 哨兵划分
    quick_sort(nums,left,pivot-1)
    quick_sort(nums,pivot+1,right)


nums = [7, 6, 8, 5, 2, 1, 3, 4, 0, 9, 10]
print(nums)
quick_sort(nums,0,len(nums)-1)
nums
```

## ✅两个栈实现一个队列  
>通过两个栈来模拟队列的先进先出（FIFO）特性。会有两个栈，一个用于入队操作（stack_cache），另一个用于出队操作（stack_out）。

具体实现方法是：
- 入队操作时，将元素推入 stack_cache 栈。
- 出队操作时，如果 stack_out 栈为空，将 stack_cache 中的所有元素弹出并推入 stack_out，然后从 stack_out 中弹出栈顶元素。如果 stack_out 不为空，直接从 stack_out 中弹出栈顶元素。

总结一下，入队操作时间复杂度为O(1)，出队操作在最坏情况下为O(n)，但摊销时间复杂度为O(1)。这种方法利用栈的后进先出（LIFO）特性来模拟队列的先进先出（FIFO）行为。  
摊销分析：不是看单次操作的最坏情况，而是看多次操作的平均时间。  

### 232. 用栈实现队列

```python
class MyQueue:
    def __init__(self):
        self.stack_cache = []
        self.stack_out = []
    def push(self, x: int) -> None:
        self.stack_cache.append(x)
    def pop(self) -> int:
        if not self.stack_out:
            while self.stack_cache:
                self.stack_out.append(self.stack_cache.pop())
        return self.stack_out.pop()
    def peek(self) -> int:
        return self.stack_cache[0] if not self.stack_out else self.stack_out[-1]
    def empty(self) -> bool:
        return not self.stack_out and not self.stack_cache
    
```

## ✅反转链表  
>面试中经常考的是反转链表的局部


```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
        
def reverseList(head):
    if not head: return None
    
    tail = head
    # 头插法（迭代版本）
    while tail.next:
        temp = tail.next
        tail.next = temp.next
        temp.next = head
        head = temp
    
    return head
```

## ✅删除单链表倒数第K个节点  
通过两个指针来实现，其中一个指针比另一个指针提前K个节点。这样，当前面的指针到达链表末尾时，后面的指针正好位于需要删除的节点的前一个位置。

总结一下，通过使用两个指针，一个比另一个提前K个节点，实现了删除单链表倒数第K个节点。  
时间复杂度为O(n)，因为我们需要遍历链表一次。  
空间复杂度为O(1)，因为我们只使用了常数级别的额外空间。


```python
def removeNthFromEnd(head, n):
    dummy = ListNode(next=head)
    slow = fast = dummy
    for _ in range(n):
        fast = fast.next
        
    while fast and fast.next:
        slow = slow.next
        fast = fast.next
    
    slow.next = slow.next.next
    
    return dummy.next
    
```


## ✅删除单链表中等于给定值val的所有节点  
>样例：给出链表 1->2->3->3->4->5->3，val = 3，你需要返回删除3之后的链表：1->2->4->5

这个问题可以通过遍历链表，并检查每个节点的值是否等于val来解决。  
可能会删除头结点，因此使用一个虚拟头节点  

总结一下，通过遍历链表并检查每个节点的值是否等于给定值 val 来删除所有匹配的节点。使用虚拟头节点可以处理删除头节点的情况。  
时间复杂度为 O(n)，因为我们需要遍历整个链表。  
空间复杂度为 O(1)，因为我们只使用了常数级别的额外空间。  


```python
class ListNode:
    def __init__(self,val=0,next= None):
        self.val = val
        self.next = next
def removeElements(head, val):
    dummy = ListNode(next=head)
    cur = dummy
    while cur.next:
        if cur.next.val == val:
            cur.next = cur.next.next
        else:
            cur = cur.next
    
    return dummy.next
```

## ✅使用链表实现栈和队列  
> 使用链表实现栈。  
> 栈的特点是后进先出（LIFO），我们可以在链表的头部插入和删除节点来实现这一点。

> 使用链表实现队列  
> 队列的特点是先进先出（FIFO），我们可以在链表的头部删除节点，在链表的尾部插入节点来实现这一点。


```python
class ListNode:
    def __init__(self,val=0,next=None):
        self.val = val
        self.next = next
    
class Stack:
    def __init__(self):
        # 始终指向头部
        self.top = None
    
    def push(self, x:int) -> None:
        node = ListNode(x)
        node.next = self.top
        self.top = node
    
    def pop(self) -> int:
        if not self.top:
            raise IndexError("pop from empty stack")
        temp = self.top
        self.top = self.top.next
        return temp.val
    def peek(self) -> int:
        if not self.top:
            raise IndexError("peek from empty stack")
        return self.top.val
    
    def empty(self) -> bool:
        return self.top is None
```

```python
class Queue:
    def __init__(self):
        # 始终指向头部和尾部
        self.top = None
        self.tail = None
    
    def enqueue(self, x:int) -> None:
        node = ListNode(x)
        if not self.tail:
            self.top = self.tail = node
        else:
            self.tail.next = node
            self.tail = node
    
    def dequeue(self) -> int:
        if not self.top:
            raise IndexError("dequeue from empty stack")
        temp = self.top
        self.top = self.top.next
        if not self.top:
            self.tail = None
        return temp.val
    def peek(self) -> int:
        if not self.top:
            raise IndexError("peek from empty stack")
        return self.top.val
    
    def empty(self) -> bool:
        return self.top is None
```

## ✅判断链表是否有环  
>给你一个链表的头节点 head ，判断链表中是否有环。

这个问题可以通过快慢指针来解决。快指针每次移动两个节点，慢指针每次移动一个节点。如果链表中存在环，快指针最终会与慢指针相遇。


```python
def hasCycle(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True
    return False
```


## ✅如果链表有环，返回链表中环的起始节点  
>找到后，slow返回头结点，和fast按照1匀速运动


```python
def detectCycle(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            slow = head
            while slow != fast:
                slow = slow.next
                fast = fast.next
            return slow
                
    return None
```

## ✅实现一个带有取最小值min方法的栈，min方法返回当前栈中的最小值  
>155. 最小栈, 这是一道参考的题。实现的栈将支持 push、pop 和 min 操作，所有操作要求都在 O(1) 时间内完成。

思路：我们可以使用一个栈，栈中元素是一个元组(val, 和min)

```python
class MinStack:

    def __init__(self):
        self.stack = []

    def push(self, val: int) -> None:
        if len(self.stack) == 0:
            self.stack.append((val,val))
        else:
            self.stack.append((val,min(val, self.stack[-1][1])))

    def pop(self) -> None:
        return self.stack.pop()[0]

    def top(self) -> int:
        return self.stack[-1][0]

    def getMin(self) -> int:
        return self.stack[-1][1]
```


## ✅使用数组实现一个栈，需要提供自动扩容和缩容功能  
我们将定义一个栈类，内部维护一个数组用于存储元素，并在需要时调整数组的大小。

总结一下，这个使用数组实现的栈通过在需要时调整数组大小来实现自动扩容和缩容功能。  
当栈满时，数组容量加倍；当栈大小小于容量的四分之一时，数组容量减半。时间复杂度为O(1)均摊，因为调整大小的操作在多次push和pop操作中摊销开来

```python
class ArrayStack:
    def __init__(self, capacity=2):
        self.stack = [None]*capacity
        self.capacity = capacity
        self.size = 0
        
    def push(self,x):
        if self.size == self.capacity:
            self._resize(2 * self.capacity)
        self.stack[self.size] = x
        self.size += 1
        
    def pop(self):
        if self.size == 0:
            raise IndexError("pop from empty stack")
        
        x = self.stack[self.size - 1]
        self.stack[self.size - 1] = None
        self.size -= 1
        if 0< self.size  < self.capacity // 4:
            self._resize(self.capacity // 2)
        
        return x
        
    def _resize(self, new_capacity):
        new_stack = [None]*new_capacity
        for i in range(self.size):
            new_stack[i] = self.stack [i]
        self.stack = new_stack
        self.capacity = new_capacity
        
    def __len__(self):
        return self.size
    
    def is_empty(self):
        return self.size == 0
    
```

```python
def test_array_stack():
    # 创建一个初始容量为2的栈
    stack = ArrayStack(2)
    print(f"初始状态: capacity={stack.capacity}, size={stack.size}")
    
    # 测试扩容：连续push元素直到触发扩容
    print("\n=== 测试扩容 ===")
    for i in range(1, 6):
        stack.push(i)
        print(f"push({i})后: capacity={stack.capacity}, size={stack.size}")
    
    # 测试正常pop
    print("\n=== 测试正常pop ===")
    for i in range(2):
        popped = stack.pop()
        print(f"pop(): {popped}, capacity={stack.capacity}, size={stack.size}")
    
    # 测试缩容：继续pop直到触发缩容
    print("\n=== 测试缩容 ===")
    while not stack.is_empty():
        popped = stack.pop()
        print(f"pop(): {popped}, capacity={stack.capacity}, size={stack.size}")
    
    # 测试边界情况
    print("\n=== 测试边界情况 ===")
    stack.push(100)
    stack.push(200)
    stack.push(300)  # 应该触发扩容
    print(f"push 3个元素后: capacity={stack.capacity}, size={stack.size}")
    
    stack.pop()
    stack.pop()  # 这两个pop后应该触发缩容
    print(f"pop 2个元素后: capacity={stack.capacity}, size={stack.size}")

# 运行测试
test_array_stack()
```

## ✅两个队列实现栈  
>通过两个队列来实现栈的后进先出（LIFO）特性。具体方法是将新元素总是加入到一个空队列中，然后将另一个队列的所有元素依次移到这个队列中，从而实现栈的效果。

总结一下，通过两个队列实现栈的算法，通过在一个空队列中加入新元素，并将另一个队列的所有元素依次移到这个队列中，从而实现栈的后进先出（LIFO）特性。  
push操作的时间复杂度为O(n)，因为需要将所有元素移动到新队列中，而pop和top操作的时间复杂度为O(1)。


```python
import collections
class MyStack:
    def __init__(self):
        self.out_queue = collections.deque()
        self.cache_queue = collections.deque()
        

    def push(self, x: int) -> None:
        self.cache_queue.append(x)
        
        while self.out_queue:
            self.cache_queue.append(self.out_queue.popleft())
            
        self.out_queue, self.cache_queue =  self.cache_queue , self.out_queue

    def pop(self) -> int:
        
        return self.out_queue.popleft() if self.out_queue else IndexError("pop from empty stack")

    def top(self) -> int:
        return self.out_queue[0] if self.out_queue else IndexError("top from empty stack")

    def empty(self) -> bool:
        return not self.out_queue
```


## ✅有效的括号序列  
>给定一个字符串所表示的括号序列，包含'( ', ')', '{', '}', '[', ']'，判定是否是有效的括号序列。  
样例：  
括号必须按照"0"顺序表示，"()[]{}"是有效的括号，但是"([)]"是无效的，")({[}]"也是无效的。  

时间复杂度：O(N), 因为要遍历一遍字符串  
空间复杂度：O(N)，因为最坏情况下需要将所有开括号放入栈中


```python
def isValid(self, s: str) -> bool:
        # 用栈这种数据结构
        stack = []
        hashmap = {')':'(','}':'{',']':'['}
        for c in s:
            if stack and c in hashmap:
                if hashmap[c] == stack[-1]:
                    stack.pop()
                else: 
                    return False  
            else:
                stack.append(c)
    
        return not stack
```

## ✅字符串查找  
>暴力法：haystack 中找needle. 这种方法的缺点在于，它遗忘了之前已经匹配成功的部分了。  
KMP (Knuth-Morris-Pratt) 算法是计算机科学中一个高效的字符串匹配算法。KMP是三个人名字的合起来  


```python
def strStr(haystack, needle):
    M,N = len(haystack) , len(needle)
    for start in range(M - N + 1):
        if haystack[start,start+N] == needle:
            return start
    return -1 
```

## ✅逆序一个栈  
>我们可以使用递归的方法。需要两个步骤：首先，通过递归的方法将栈顶元素取出，然后递归逆序剩余的栈。其次，将取出的元素插入到逆序后的栈底。

两个递归函数的栈的物理地址是一个吗？  
好像是一个


```python
def insertAtBottom(stack,value):
    if not stack:
        stack.append(value)
        return
    top = stack.pop()
    insertAtBottom(stack, top)
    stack.append(top)

def reverseStack(stack):
    # 归
    if not stack: return 
    
    # 递
    top = stack.pop()
    reverseStack(stack)
    insertAtBottom(stack, top)
```

## ✅用一个栈实现对另一个栈的排序  
>面试题中叫做双栈排序

通过一个辅助栈来对另一个栈进行排序。  
具体来说，我们将使用一个辅助栈来帮助排序原栈中的元素。  
这个算法的思想是将原栈中的元素逐个弹出，并按顺序插入到辅助栈中，从而实现排序。  



总结一下，这个算法通过一个辅助栈来对原栈中的元素进行排序。  
时间复杂度为O(n²)，因为每个元素在最坏情况下需要移动n次。  
空间复杂度为O(n)，因为我们使用了一个额外的栈来存储排序元素。  


```python
def stack_sort(nums):
    tmp = []
    while nums:
        curr = nums.pop()
        # 保证临时栈的栈低是最小值
        while tmp and tmp[-1] > curr:
            nums.append(tmp.pop())
        tmp.append(curr)
    # while tmp:
    #     nums.append(tmp.pop())
    return tmp

nums = [7, 6, 8, 5, 2, 1, 3, 4, 0, 9, 10]
stack_sort(nums)
# nums
```

## ✅删除无序链表中值重复出现的节点  
>方法1：使用哈希集合seen = set(), 遇到在哈希集合中的节点就删除。    
时间复杂度：O(N)  
空间复杂度：O(N)  

>方法2：利用指针比较。当cur.next.val 等于 cur.val 节点就删除。  
时间复杂度：O(N^2), 因为每个节点都要与后序节点比较  
空间复杂度：O(1) 


```python
# 无序链表
def deleteDuplicates(head):

    seen = set()
    dummy = ListNode(next=head)
    curr = dummy
    while curr.next:
        if curr.next.val in seen:
            curr.next = curr.next.next
        else:
            seen.add(curr.next.val)
            curr = curr.next
    
    return dummy.next
```

## ✅删除排序链表中的重复元素  
>由于链表是有序的，我们可以通过遍历链表并比较相邻节点的值来删除重复的节点  

总结一下，这个算法通过遍历排序链表并比较相邻节点的值来删除重复节点。    
由于链表是有序的，相同值的节点会相邻出现，这使得我们可以高效地删除重复节点。    
时间复杂度为O(n)，因为我们需要遍历整个链表一次。  
空间复杂度为O(1)，因为我们只使用了常数级别的额外空间。  

```python
# 有序链表
def deleteDuplicates(head):
    if not head or not head.next: return head

    cur = head
    while cur.next:
        if cur.next.val == cur.val:
            cur.next = cur.next.next
        else:
            cur = cur.next
    
    return head
```

## ✅向有序环形链表中加入新的节点  
>环形链表的特点是最后一个节点指向头节点，使链表形成一个环。我们需要在适当的位置插入新节点，保持链表的有序性。

总结一下，这个算法通过遍历有序环形链表，找到适当的位置插入新节点，保持链表的有序性。  
时间复杂度为O(n)，因为在最坏情况下，我们需要遍历整个链表一次。  
空间复杂度为O(1)，因为我们只使用了常数级别的额外空间。  


```python
class ListNode:
    def __init__(self,val=0,next= None):
        self.val = val
        self.next = next

def insert(head, insertVal):
    new_node = ListNode(insertVal)
    
    if not head:
        new_node.next = new_node
        return new_node
    
    prev, curr = head,head.next
    
    to_insert = False
    
    while True:
        # 
        if prev.val <= insertVal <= curr.val:
            to_insert = True
        elif prev.val > curr.val:
            if insertVal >= prev.val or prev.val <= curr.val:
                to_insert = True
        if to_insert:
            prev.next = new_node
            new_node.next = curr
            return head
        
        prev, curr = curr,curr.next
        if prev == head:
            break
        
    prev.next = new_node
    new_node.next = curr
    return head
        
```

## ✅判断两个字符串是否互为变形词，也叫字母异位词  
>两个字符串互为变形词的条件是，它们包含的字符及每个字符的出现次数相同。我们可以通过计数每个字符的出现次数来解决这个问题。

总结一下，通过使用字典记录每个字符的出现次数，可以高效地判断两个字符串是否互为变形词。  
时间复杂度为O(n)，因为我们需要遍历两个字符串一次。  
空间复杂度为O(1)，因为字母表的大小是固定的，不会随输入字符串的长度而变化。  


```python
import collections

def isAnagram(s: str, t: str) -> bool:
    if len(s) != len(t): return False
    
    count = collections.defaultdict(int)
    
    # 遍历第一个，增加
    for char in s:
        count[char] += 1

    # 遍历第二个，减少 
    for char in t:
        count[char] -= 1
        if count[char] < 0:
            return False
        
            
    return sum(count.values()) == 0

s = "anagram"
t = "nagaram"
isAnagram(s,t)
```

## ✅二分查找  
>二分查找算法在有序数组中查找目标值，时间复杂度为O(log n)。通过不断将查找范围减半来实现这一目标。


```python
def search(nums, target):
    left, right = 0, len(nums) - 1
    while left <= right:
        mid = left + (right - left) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid + 1 
        else:
            right = mid - 1
    return -1
```

## ✅二分查找 follow up  
>给定一个排序的整数数组（升序）和一个要查找的整数 target，用 O(log N) 的时间找到 target 最后一次出现的下标（从 0 开始）。  
如果 target 不存在于数组中，返回 -1。  
样例：  
在数组 [1, 2, 3, 3, 4, 5, 10] 中二分查找 3，返回 3。  


总结一下，这个算法通过二分查找在 O(log N) 时间内找到 target 最后一次出现的下标。  
我们初始化左右指针来定义查找范围，并在查找过程中不断调整左右指针。  
如果找到 target，则记录当前位置并继续向右查找，直到查找范围内没有 target。  


```python
# 右边界下标，这个改进的好
def search(nums, target):
    left, right = 0, len(nums) - 1
    result = -1
    while left <= right:
        mid = left + (right - left) // 2
        if nums[mid] == target:
            result =  mid
            left = mid + 1 
        elif nums[mid] < target:
            left = mid + 1 
        else:
            right = mid - 1
    return result
```
