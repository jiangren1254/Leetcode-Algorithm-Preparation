## ✅选择排序、插入排序、冒泡排序  
>这三个的时间复杂度都是O(N^2)


```python
# 选择排序思想：每轮从未排序区间选择最小的元素，将其放到已排序区间的末尾。
# 时间复杂度：O(N^2)
# 算法特点：排序时间和初始顺序无关，即使初始已经排好序了，也要耗费O(N^2)的时间
def selection_sort(nums):
    N = len(nums)
    for i in range(N):
        # 选择未排序区间中最小的元素
        min_idx = i
        for j in range(i+1,N):
            if nums[j] < nums[min_idx]:
                min_idx = j
        # 将其放到已排序区间的末尾
        nums[i], nums[min_idx] = nums[min_idx], nums[i]
    return nums

nums = [7, 6, 8, 5, 2, 1, 3, 4, 0, 9, 10]
print(nums)
selection_sort(nums)
```

```python
# 冒泡排序（bubble sort）通过连续地比较与交换相邻元素实现排序。这个过程就像气泡从底部(nums 左边)升到顶部(nums 右边)一样，因此得名冒泡排序。
# 时间复杂度：O(N^2)
def bubble_sort(nums):
    N = len(nums)
    for i in range(N):
        # 是否发生了交换相邻元素
        isSwapped = False
        for j in range(0, N-i-1):
            if nums[j] > nums[j+1]:
                # 交换元素
                nums[j], nums[j+1] = nums[j+1], nums[j]
                isSwapped = True
        # 如果没有交换，数组已经有序，退出循环
        if not isSwapped:
            break
        
    return nums

nums = [7, 6, 8, 5, 2, 1, 3, 4, 0, 9, 10]
print(nums)
bubble_sort(nums)
```

```python
# 插入排序思想：就像打牌时整理牌，小的插到左边，大的插到右边
# 算法特点：最坏情况下时间复杂度O(N^2)
# 依赖于初始排列顺序，特别适合于部分有序的数组
# 完全有序的情况下，复杂度能达到O(N)，但逆序时效率降到O(N^2)
def insertion_sort(nums: list[int]):
    """ 插入排序"""
    # 外循环：已排序区间为 [0, i-1]
    for i in range(1, len(nums)):
        base = nums[i]
        j = i - 1
        # 内循环：将 base 插入到已排序区间 [0, i-1] 中的正确位置
        while j >= 0 and base < nums[j]:
            nums[j + 1] = nums[j]
            # 将 nums[j] 向右移动一位
            j -= 1
        nums[j + 1] = base  # 将 base 赋值到正确位置
        
nums = [7, 6, 8, 5, 2, 1, 3, 4, 0, 9, 10]
print(nums)
insertion_sort(nums)
nums
```

## ✅归并排序、快速排序  
>这两个都是基于分治的算法


```python
# 归并排序是一种基于分治法的排序算法，它将数组递归地分成两个子数组，分别排序后再合并。
# 时间复杂度为O(n log n)，因为数组每次都被二分，合并的时间复杂度为O(n)。
# 空间复杂度为O(n)，因为需要额外的空间来存储临时数组。

def findMid(nums):
    return len(nums) // 2

def mergeTwoNums(nums1, nums2):
    i,j = 0,0
    result = []
    while i < len(nums1) and j < len(nums2):
        if nums1[i] <= nums2[j]:  # 使用 <= 保证稳定性
            result.append(nums1[i])
            i+=1
        else:
            result.append(nums2[j])
            j+=1
        
    result += nums1[i:] if i < len(nums1) else nums2[j:]
        
    return result

def merge_sort(nums):
    if len(nums)<2:
        return nums.copy()
    
    mid = findMid(nums)
    left_res = merge_sort(nums[:mid])
    right_res = merge_sort(nums[mid:])
    
    return mergeTwoNums(left_res, right_res)


nums = [7, 6, 8, 5, 2, 1, 3, 4, 0, 9, 10]
# print(findMid(nums))
print(nums)
merge_sort(nums)
# nums
```

```python
# 快速排序也是一种基于分治法的排序算法。它通过选择一个“枢纽”元素，将数组分成两部分，其中一部分元素小于枢纽，另一部分大于枢纽，然后递归地对两部分排序。
# 时间复杂度为O(n log n)（平均情况），因为每次分区大约将数组分成两半。最坏情况下（如已排序数组），时间复杂度为O(n²)
# 空间复杂度为O(log n)，因为递归调用栈的深度为log n
def partition(nums, left, right):
    # 哨兵划分
    pivot_idx = left #选择最左边的元素称为枢纽
    pivot = nums[pivot_idx] 
    i , j= left, right
    # 从右往左找出第一个小于pivot的数
    # 从左往右找出第一个大于pivot的数
    while i<j:
        while i <j and nums[j] >= pivot:
            j -= 1
        while i < j and nums[i] <= pivot:
            i += 1
        if i<j:
            nums[i],nums[j] = nums[j], nums[i]
    # 全都交换完毕，此时i=j
    nums[i],nums[pivot_idx] = nums[pivot_idx],nums[i]
    
    return i

def quick_sort(nums,left,right):
    if left >= right: return
    
    pivot = partition(nums, left, right)  # 哨兵划分
    quick_sort(nums,left,pivot-1)
    quick_sort(nums,pivot+1,right)


nums = [7, 6, 8, 5, 2, 1, 3, 4, 0, 9, 10]
print(nums)
quick_sort(nums,0,len(nums)-1)
nums
```

## ✅两个栈实现一个队列  
>通过两个栈来模拟队列的先进先出（FIFO）特性。会有两个栈，一个用于入队操作（stack_cache），另一个用于出队操作（stack_out）。

具体实现方法是：
- 入队操作时，将元素推入 stack_cache 栈。
- 出队操作时，如果 stack_out 栈为空，将 stack_cache 中的所有元素弹出并推入 stack_out，然后从 stack_out 中弹出栈顶元素。如果 stack_out 不为空，直接从 stack_out 中弹出栈顶元素。

总结一下，入队操作时间复杂度为O(1)，出队操作在最坏情况下为O(n)，但摊销时间复杂度为O(1)。这种方法利用栈的后进先出（LIFO）特性来模拟队列的先进先出（FIFO）行为。  
摊销分析：不是看单次操作的最坏情况，而是看多次操作的平均时间。  

### 232. 用栈实现队列
>仅使用两个栈实现先入先出队列。队列应当支持一般队列支持的所有操作（push、pop、peek、empty）

相似的题：
- LCR 125. 图书整理 II
- 225. 用队列实现栈(暂时没做)

```python
class MyQueue:
    def __init__(self):
        self.stack_cache = []
        self.stack_out = []
    def push(self, x: int) -> None:
        self.stack_cache.append(x)
    def pop(self) -> int:
        if not self.stack_out:
            while self.stack_cache:
                self.stack_out.append(self.stack_cache.pop())
        return self.stack_out.pop()
    def peek(self) -> int:
        return self.stack_cache[0] if not self.stack_out else self.stack_out[-1]
    def empty(self) -> bool:
        return not self.stack_out and not self.stack_cache
    
```

## ✅反转链表  
>面试中经常考的是反转链表的局部


```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
        
def reverseList(head):
    if not head: return None
    
    tail = head
    # 头插法（迭代版本）
    while tail.next:
        temp = tail.next
        tail.next = temp.next
        temp.next = head
        head = temp
    
    return head
```

## ✅删除单链表倒数第K个节点  
通过两个指针来实现，其中一个指针比另一个指针提前K个节点。这样，当前面的指针到达链表末尾时，后面的指针正好位于需要删除的节点的前一个位置。

总结一下，通过使用两个指针，一个比另一个提前K个节点，实现了删除单链表倒数第K个节点。  
时间复杂度为O(n)，因为我们需要遍历链表一次。  
空间复杂度为O(1)，因为我们只使用了常数级别的额外空间。


```python
def removeNthFromEnd(head, n):
    dummy = ListNode(next=head)
    slow = fast = dummy
    for _ in range(n):
        fast = fast.next
        
    while fast and fast.next:
        slow = slow.next
        fast = fast.next
    
    slow.next = slow.next.next
    
    return dummy.next
    
```


## ✅删除单链表中等于给定值val的所有节点  
>样例：给出链表 1->2->3->3->4->5->3，val = 3，你需要返回删除3之后的链表：1->2->4->5

这个问题可以通过遍历链表，并检查每个节点的值是否等于val来解决。  
可能会删除头结点，因此使用一个虚拟头节点  

总结一下，通过遍历链表并检查每个节点的值是否等于给定值 val 来删除所有匹配的节点。使用虚拟头节点可以处理删除头节点的情况。  
时间复杂度为 O(n)，因为我们需要遍历整个链表。  
空间复杂度为 O(1)，因为我们只使用了常数级别的额外空间。  


```python
class ListNode:
    def __init__(self,val=0,next= None):
        self.val = val
        self.next = next
def removeElements(head, val):
    dummy = ListNode(next=head)
    cur = dummy
    while cur.next:
        if cur.next.val == val:
            cur.next = cur.next.next
        else:
            cur = cur.next
    
    return dummy.next
```
