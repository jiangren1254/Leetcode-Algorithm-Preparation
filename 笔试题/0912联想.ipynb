- 模拟的思路：先分快，再求最大值。如何分？穷举不可行。
- 先排序, 二分
- 前缀和。碰到均值切分一下. + 二分

```python
import bisect
def main():
    # n = int(input())
    # array = list(map(int, input().split()))
    n = 10
    array = [2, 5, 1, 4, 7, 3, 6, 2, 5, 1]
    pre = [0]*(n+1)
    for i in range(n):
        pre[i+1] = array[i] + pre[i]
    print(pre)
    
    total = pre[-1]
    result = float('inf')
    i = bisect.bisect_left(pre,total // 3)
    # print(i)

    # for left in range(1, i):  # 这里可以优化一下时间
    for cut1 in [i, i-1]:
        if 1<= cut1 <= n-2:
            S1 = pre[cut1]
            rest = total -S1
            target = S1 + rest // 2
            j = bisect.bisect_left(pre, target, cut1+1,n)
            for cut2 in [j,j-1]:  # 找出两者中的最优
                if cut1<cut2<n:
                    S2 = pre[cut2] - pre[cut1]
                    S3 = total - pre[cut2]
                    S= max(S1,S2,S3)
                    result = min(result, S)

    return result
        
    

main()
```
