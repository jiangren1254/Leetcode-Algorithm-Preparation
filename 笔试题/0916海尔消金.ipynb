### 未出现的字符总和
> 求解未出现的字符总和

python的集合遍历：因为集合元素是无序的，所以只能通过for x in s:  print(x), 直接遍历  
python 中求解字符的ASCII。ord('A') 《》 chr(65)

```python
TOTAL = 0
for i in range(65,91):
    TOTAL += i
TOTAL,TOTAL - 65
```

学堂分组

```python
# nums = [100, 110, 120, 140, 160, 190,200,250]
# nums = [100,100,100]
nums = [60,100]
n = len(nums)
nums.sort()
nums
```

- 分组，最少组。回溯行吗？回溯可以统计每组的元素。
- 思路2：就三个选择，3，2，1，。贪心做法，最左边是
- 滑动窗口+贪心。反向遍历

```python
left=0
cnt = 0
while left < n:
    step = 2
    while step > 0:
        # print(step)
        right = min(n-1,left+step)
        if step == 2:
            if nums[right] - nums[left] <=10:
                print(left,right)
                left = right+1
                cnt += 1
                break
        elif step == 1:
            if nums[right] - nums[left] <=20:
                print(left,right)
                left = right+1
                cnt += 1
                break
        # 变化
        step -= 1
    # post
    if step == 0:
        print(left)
        cnt += 1
        left+=1
print(cnt)
```
