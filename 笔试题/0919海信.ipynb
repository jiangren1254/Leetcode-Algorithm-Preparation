### 1. 数组统一问题
>请判断是否存在一个整数 x ，使得经过若干次操作后，数组中的所有元素都等于 x 。

- 思路：

```python
def main():
    # T = int(input())
    T = 1
    results = []
    for _ in range(T):
        n, m = map(int, input().split())
        nums = list(map(int, input().split()))
        d = [(i+1) % m for i in range(n)]
        fixed_vals = []  # d[i] == 0 对应的nums[i] 不变
        for i in range(n):
            if d[i] == 0:
                fixed_vals.append(nums[i])
        # 所有固定值必须相等
        x = fixed_vals[0]
        for val in fixed_vals:
            if val != x:
                results.append("NO")
                break
        # 因为m <= n一定有固定数， 检查所有非固定位置是否能变为 x
        valid = True
        for i in range(n):
            if d[i] != 0:
                if (x - nums[i]) % d[i] != 0:
                    valid = False
                    break
        results.append("YES" if valid else "NO")
        
    for res in results:
        print(res)
        
main()


# 2
# 3 3
# 1 6 8
# 4 3
# 2 6 7 5
```

### 2.
贪心解法

```python
def main():
    T = int(input())
    for _ in range(T):
        n, m, x, y = map(int, input().split())
        prices = list(map(int, input().split()))
        
        sold = 0
        total_income = 0
        
        for price in prices:
            if price >= y and sold < n:
                total_income += price
                sold += 1
        
        profit = total_income - n * x
        print(profit)

main()
```
