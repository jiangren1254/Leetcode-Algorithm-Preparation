```python
def main():
    n = int(input())
    a = list(map(int, input().split()))

    if n == 1:
        print("Yes" if a[0] == 1 else "No")
    else:
        possible = False
        
        # 枚举 x[0] 的两种情况
        for x0 in range(2):
            x = [0] * n
            x[0] = x0
            x[1] = 1 ^ a[0]  # 根据灯0的约束确定 x[1]
            
            # 递推计算后续的 x 值
            for i in range(1, n - 1):
                x[i + 1] = 1 ^ a[i] ^ x[i - 1]
            
            # 检查最后一个灯的约束是否满足
            if (a[n - 1] ^ x[n - 2]) == 1:
                possible = True
                break
        
        print("Yes" if possible else "No")
main()
```

```python
import sys
MOD = 10**9 + 7

def main():    
    input = sys.stdin.buffer.readline
    q = int(input().strip())
    
    cnt = [0] * 30   # 每一位的子数组数量
    ans = 0
    pow2 = [1 << k for k in range(30)]
    out = []
    
    for _ in range(q):
        op = input().split()
        if op[0] == b"1":   # 注意 buffer 输入是字节串
            x = int(op[1])
            for k in range(30):
                if (x >> k) & 1:
                    cnt[k] += 1
                else:
                    cnt[k] = 0
                ans += cnt[k] * pow2[k]
            ans %= MOD
        else:  # op[0] == "2"
            out.append(str(ans % MOD))
    
    sys.stdout.write("\n".join(out))

main()
```
