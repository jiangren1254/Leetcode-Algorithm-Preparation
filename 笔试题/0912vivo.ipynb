```python
import collections
def sort(tasks, events, recommend_events):
    events2tasks =  {events[i]:tasks[i] for i in range(len(tasks))}
    print(events2tasks)
    results = []
    seen = set()
    
    for i in range(len(recommend_events)):
        ev = recommend_events[i]
        if ev in events2tasks:
            task = events2tasks[ev]
            if task not in seen:
                results.append(task)
                seen.add(task)
    
    results2 = []
    for i in range(len(tasks)):
        task = tasks[i]
        if tasks[i] not in seen:
            results2.append(task)
            seen.add(task)
    print(results2 + results) 
    
    return (results2 + results)[:10]

tasks = [3,1,5,9,2]
events = ["f","g","c","d","a"]
recommend_events = ["a","b","c","d"]
sort(tasks, events, recommend_events)
```

- 暴力解法：每个元素都删除一遍，计算最小波动值

```python
def maxWaveValue(n: int, nums) -> int:
    
    totals = []
    for i in range(n):
        new_nums = []
        for j in range(n):
            if j != i:
                new_nums.append(nums[j])
        # print(new_nums)
        totals.append(new_nums)
    print(totals)
    
    max_val = float('-inf')
    for i in range(len(totals)):
        new_nums = totals[i]
        new_nums.sort()
        results = []
        def backtrack(start=0, path=[]):
            if len(path) == 2:
                results.append(abs(path[0]- path[1]))
                return

            for i in range(start, len(new_nums)):
                if i > start and new_nums[i] == new_nums[i-1]:
                    continue
                path.append(new_nums[i])
                backtrack(i+1)
                path.pop()
        
        results = []
        backtrack()
        max_val = max(max_val,min(results))
        print(min(results))  
        
    return max_val

# nums = [4,2,8,1]
nums = [2,8,1]
nums = [1,2,1]
# nums = [5,5,5]
n = len(nums)

maxWaveValue(n,nums)
```

三维DP

```python
from typing import List
def findCheapestPrice(n: int, src: int, dst: int, k: int, c: int, t: int, wormholes: List[List[int]]) -> int:
    graph = [[] for _ in range(n)]
    # print(graph)
    # print(wormholes)
    for i in range(len(wormholes)):
        u, v, cost, energy, time = wormholes[i]
        graph[u].append((v, cost, energy, time))
    # print(graph)
    
    # 自定义函数
    def add_state(state, new_state):
        return state, new_state
    
    max_edge = k - 1 # 中转次数
    dp = [[[] for _ in range(n)] for _ in range(k)]
    dp[0][src] = [(0,0,0)]
    print(dp)
    for e in range(max_edge):
        # print(i)
        for v in range(n):
            for state in dp[e][v]:
                for edge in graph[v]:
                    nxetV, cost, energy, time = edge
                    new_cost = state[0] + cost
                    new_energy = state[1]+ energy
                    new_time = state[2]+ time
                    if new_energy <= c and new_time <= t:
                        new_state = (new_cost, new_energy, new_time)
                        dp[e+1][nxetV] = add_state(dp[e+1][nxetV], new_state)
        
    return -1
        

n, src, dst, k, c, t, wormholes = 3,0,2,1,10,10,[[0,1,5,2,3], [1,2,3,1,4], [0,2,10,5,5]]
findCheapestPrice(n, src, dst, k, c, t, wormholes)
```
