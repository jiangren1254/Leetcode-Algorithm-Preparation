### 1、数组中逆序对的总数

> 找出数组中的逆序对的总数

思路： 原题是：LCR 170. 交易逆序对的总数，使用归并排序的模版。在并的过程中记下来逆序对。甚至最后对数组完成了排序

```python
def mergeSort(record, tmp, l, r):
    if l >= r:
        return 0

    mid = (l + r) // 2
    inv_count = mergeSort(record, tmp, l, mid) + mergeSort(record, tmp, mid + 1, r)
    # 开始 并
    i, j, pos = l, mid + 1, l
    while i <= mid and j <= r:
        if record[i] <= record[j]:
            tmp[pos] = record[i]
            i += 1
            inv_count += (j - (mid + 1))  # 这里算贡献
        else:
            tmp[pos] = record[j]
            j += 1
        pos += 1
    # 把后面的拼上
    for k in range(i, mid + 1):
        tmp[pos] = record[k]
        inv_count += (j - (mid + 1))
        pos += 1
    for k in range(j, r + 1):
        tmp[pos] = record[k]
        pos += 1
    # 辅助数组tmp 拷贝回原数组。
    record[l:r+1] = tmp[l:r+1]
    return inv_count

def main():
    array = list(map(int, input().split()))
    n = len(array)
    tmp = [0] * n
    return mergeSort(array, tmp, 0, n - 1)   

main()
```

### 2.接水问题
> n 名同学,m个水龙头，所有同学都接完水需要多少秒？

思路：这是一个典型的模拟问题，下面是优先队列(堆)的实现。水龙头大小就是堆的大小。答案就是最后heap中的最大值

```python
import heapq

def main():
    # n, m = map(int, input().split())
    # nums = list(map(int, input().split()))
    n,m = 6, 3 
    nums = [4, 4, 1, 2, 1, 5]
    heap = [0] * m
    for i in range(n):
        min_time = heapq.heappop(heap)
        heapq.heappush(heap, min_time + nums[i])

    return max(heap)

main()
```
