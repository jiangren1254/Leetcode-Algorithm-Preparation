### 394. 字符串解码
> 给定一个经过编码的字符串，返回它解码后的字符串。

- 思路：因为字符串中只有数字字符，字母字符，和[,]这四种字符。所以正向遍历的时候，分别对四种情况做不同处理。下面是适用两个栈实现的代码。

```python
def decodeString(s: str) -> str:
    stack_num, stack_str = [], ['']
    
    num = 0
    for i in range(len(s)):
        char = s[i]
        if '0' <= char <='9':
            num = num *10 + int(char)
        elif char == '[':
            stack_num.append(num)
            num = 0
            stack_str.append('')
        elif char == ']':
            tmp = stack_str.pop()
            stack_str[-1] += stack_num.pop() * tmp
        else:
            stack_str[-1] += char
    
    # print(stack_str)
    # print(stack_num)
    return stack_str[0]

s = "3[a2[c]]"
decodeString(s)
```
