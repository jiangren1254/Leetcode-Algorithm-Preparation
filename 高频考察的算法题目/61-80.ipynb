### 394. 字符串解码
> 给定一个经过编码的字符串，返回它解码后的字符串。

- 思路：因为字符串中只有数字字符，字母字符，和[,]这四种字符。所以正向遍历的时候，分别对四种情况做不同处理。下面是适用两个栈实现的代码。

```python
def decodeString(s: str) -> str:
    stack_num, stack_str = [], ['']
    
    num = 0
    for i in range(len(s)):
        char = s[i]
        if '0' <= char <='9':
            num = num *10 + int(char)
        elif char == '[':
            stack_num.append(num)
            num = 0
            stack_str.append('')
        elif char == ']':
            tmp = stack_str.pop()
            stack_str[-1] += stack_num.pop() * tmp
        else:
            stack_str[-1] += char
    
    # print(stack_str)
    # print(stack_num)
    return stack_str[0]

s = "3[a2[c]]"
decodeString(s)
```

### 329. 矩阵中的最长递增路径
> 给定一个 m x n 整数矩阵 matrix ，找出其中 最长递增路径 的长度。只能从上下左右四个方向移动。

- 思路：使用记忆化深度优先搜索，当访问到一个单元格 (i,j) 时，如果 memo[i][j]=0，说明该单元格的结果已经计算过，则直接从缓存中读取结果，如果 memo[i][j]=0，说明该单元格的结果尚未被计算过，则进行搜索，并将计算得到的结果存入缓存中。
  



```python
from functools import lru_cache
from typing import List
DIRS = [(-1, 0), (1, 0), (0, -1), (0, 1)]
def longestIncreasingPath(matrix: List[List[int]]) -> int:
    if not matrix:
        return 0
    
    @lru_cache(None)
    def dfs(row: int, column: int) -> int:
        best = 1
        for dx, dy in DIRS:
            newRow, newColumn = row + dx, column + dy
            if 0 <= newRow < rows and 0 <= newColumn < columns and matrix[newRow][newColumn] > matrix[row][column]:
                best = max(best, dfs(newRow, newColumn) + 1)
        return best

    ans = 0
    rows, columns = len(matrix), len(matrix[0])
    for i in range(rows):
        for j in range(columns):
            ans = max(ans, dfs(i, j))
    # print(f"\n缓存统计: {dfs.cache_info()}")  
    return ans

matrix = [[9,9,4],[6,6,8],[2,1,1]]
longestIncreasingPath(matrix)

```

### 209. 长度最小的子数组
> 给定一个含有 n 个正整数的数组和一个正整数 target 。找出该数组中满足其总和大于等于 target 的长度最小的 子数组,返回其长度。

- 思路：滑动窗口法，这题和76. 最小覆盖子串一个解题模版。高级前缀和，带加减的。直接过了，啊哈哈哈哈

```python
import collections
def minSubArrayLen(target: int, nums: List[int]) -> int:
    l,r = 0,0
    ans = float('inf')
    prefix = 0
    while r<len(nums):
        prefix += nums[r]  # c是即将加入窗口的字符
        r += 1
        while prefix >= target:
            # 根据题意计算结果
            ans = min(ans,r-l)             
            # 进行窗口数据一系列更新
            prefix -= nums[l]
            l += 1

    return ans if ans!= float('inf') else 0
    
target = 4
nums = [1,4,4]
minSubArrayLen(target, nums)
```

### 224. 基本计算器
> 给你一个字符串表达式 s ，请你实现一个基本计算器来计算并返回它的值。

- 思路：这题是 150. 逆波兰表达式求值 的演化而来。字符串解码和这题也有点像。用队列试试

```python
def calculate(s: str) -> int:
    ops = [1]
    sign = 1

    ret = 0
    n = len(s)
    i = 0
    while i < n:
        if s[i] == ' ':
            i += 1
        elif s[i] == '+':
            sign = ops[-1]
            i += 1
        elif s[i] == '-':
            sign = -ops[-1]
            i += 1
        elif s[i] == '(':
            ops.append(sign)
            i += 1
        elif s[i] == ')':
            ops.pop()
            i += 1
        else:
            num = 0
            while i < n and s[i].isdigit():
                num = num * 10 + ord(s[i]) - ord('0')
                i += 1
            ret += num * sign
    return ret


s = "1 + 1"
s = " 2-1 + 2 "
# s = "(1+(4+5+2)-3)+(6+8)"
# s = "1-(     -2)"
# s= "- (3 + (4 + 5))"
calculate(s)
```

### 40. 组合总和 II
> 给定一个候选人编号的集合 candidates 和一个目标数 target ，找出 candidates 中所有可以使数字和为 target 的组合。  
> candidates 中的每个数字在每个组合中只能使用 一次 。

- 思路：回溯的子集和解法。in_path,控制不重复选

```python
def combinationSum2(candidates: List[int], target: int) -> List[List[int]]:
    candidates.sort()
    N = len(candidates)
    ans = []
    in_path = [False]*N
    def backtrack(start=0, path=[], path_sum = 0):
        if path_sum == target:
            ans.append(path.copy())
            return
        if path_sum > target:
            return
        last = None
        for i in range(start, N):
            if candidates[i] == last:
                continue
            if not in_path[i]:
                # 尝试
                path.append(candidates[i])
                path_sum += candidates[i]
                in_path[i] = True
                # 进行下一轮选择
                backtrack(i,path,path_sum)
                # 回退
                in_path[i] = False
                last = path.pop()
                path_sum -= last

    backtrack()
    return ans
candidates = [10,1,2,7,6,1,5]
target = 8
combinationSum2(candidates, target)
```

### 128. 最长连续序列
> 给定一个未排序的整数数组 nums ，找出数字连续的最长序列（不要求序列元素在原数组中连续）的长度。

- 思路：并查集

```python
def longestConsecutive(nums: List[int]) -> int:
    parent, length  = {}, {}

    def find(x):
        if parent[parent[x]] != parent[x]:
            parent[x] = find(parent[x])
        return parent[x]
    
    def union(x,y):
        px, py = find(x), find(y)
        if px == py:
            return
        # union by size
        if length[px] > length[py]:
            parent[py] = px
            length[px] += length[py]
        else:
            parent[px] = py
            length[py] += length[px]
        return
    # 每个都不联通
    for num in nums:
        parent[num] = num
        length[num] = 1
    # print(parent)
    # print(length)
    max_length = 0
    for num in nums:
        # 合并的条件是？
        if num+1 in parent:
            union(num+1,num)
        if num-1 in parent:
            union(num-1,num)
        max_length = max(max_length, length[parent[num]])
    return max_length

nums = [100,4,200,1,3,2]
longestConsecutive(nums)
```

### 84. 柱状图中最大的矩形
> 给定 n 个非负整数，用来表示柱状图中各个柱子的高度。每个柱子彼此相邻，且宽度为 1。求在该柱状图中，能够勾勒出来的矩形的最大面积。

- 思路：本题是hard, 用的是最小栈。一次正向遍历。代码很短。

```python
from typing import List
def largestRectangleArea(heights: List[int]) -> int:
    minStack = [-1]
    ans = 0
    for i in range(len(heights)):
        while len(minStack)>1 and heights[i] < heights[minStack[-1]]:
            h_idx = minStack.pop()
            ans = max(ans,(i-minStack[-1]-1) * heights[h_idx])
        minStack.append(i)
    # post
    while len(minStack)>1:
        h_idx = minStack.pop()
        ans = max(ans,(len(heights)-minStack[-1]-1) * heights[h_idx])
             
    return ans

heights = [2,1,5,6,2,3]
largestRectangleArea(heights)
```
