[回溯](./advanced_algorithm/backtrack.md)  

回溯法（backtrack）常用于遍历列表所有子集，是 DFS 深度搜索一种，一般用于全排列，穷尽所有可能，遍历的过程实际上是一个决策树的遍历过程。  
时间复杂度一般 O(N!)，它不像动态规划存在重叠子问题可以优化，回溯算法就是纯暴力穷举，复杂度一般都很高。

## 模版
```python
result = []
func backtrack(选择列表,路径):
    if 满足结束条件:
        result.add(路径)
        return
    for 选择 in 选择列表:
        做选择
        backtrack(选择列表,路径)
        撤销选择

```
核心就是从选择列表里做一个选择，然后一直递归往下搜索答案，如果遇到路径不通，就返回来撤销这次选择。

### 78. 子集
>给定一组不含重复元素的整数数组 nums，返回该数组所有可能的子集（幂集）。

思路：全排列问题，用回溯法

Q 为什么用 path.copy()?  
.copy() 会创建一个新的列表对象，内容是当前 path 的快照。  
这样即使后面 path 继续被修改，result 里存储的那一份不会受到影响。

````python
def subsets(nums):
    N = len(nums)
    result = []
    def backtrack(start,k, path=[]):
        if len(path) == k:
            result.append(path.copy())
            return
        for i in range(start, N):
            # 尝试
            path.append(nums[i])
            # 进行下一轮选择
            backtrack(i+1,k)
            # 回退
            path.pop()
    for k in range(N+1):
        backtrack(0,k)
    return result
nums = [1,2,3]
subsets(nums)
````

### 90. 子集 II
>给定一个可能包含重复元素的整数数组 nums，返回该数组所有可能的子集（幂集）。说明：解集不能包含重复的子集。

思路：在进行选择的时候不选择同一个元素

````python
def subsetsWithDup(nums):
    nums.sort()  # 这是子集问题不重复的关键
    N = len(nums)
    result = []
    def backtrack(start, k, path=[]):
        if len(path) == k:
            result.append(path.copy())
            return
        last = None
        for i in range(start,N):
            if nums[i] != last:  # 剪枝
                path.append(nums[i])
                backtrack(i+1,k)
                last = path.pop()
    for k in range(N + 1):
        backtrack(0, k)
        
    return result
nums = [1,2,2,2]

subsetsWithDup(nums)
````

### 46. 全排列
>给定一个不含重复数字的数组 nums ，返回其 所有可能的全排列。

- 思路 1：需要记录已经选择过的元素，满足条件的结果才进行返回，需要额外 O(n) 的空间


````python
def permute(nums):
    # nums.sort()
    N = len(nums)
    result = []
    in_path = [False]*N
    def backtrack(path=[]):
        if len(path) == N:
            result.append(path.copy())
            return
        
        for i in range(N):
            if not in_path[i]:
                path.append(nums[i])
                in_path[i] = True
                backtrack()
                path.pop()
                in_path[i] = False
    backtrack()
        
    return result
nums = [1,2,3]
permute(nums)
````

### 47. 全排列 II
>给定一个可包含重复数字的序列，返回所有不重复的全排列。

- 思路：用上子集II的last思路就好了，last = path.pop()。代码主体和全排列 I一样 

注意此题（貌似）无法使用上题的思路 2，因为交换操作会打乱排序。

````python
def permuteUnique(nums):
    # nums.sort()
    N = len(nums)
    result = []
    in_path = [False]*N
    def backtrack(path=[]):
        if len(path) == N:
            result.append(path.copy())
            return
        last = None
        for i in range(N):
            if not in_path[i] and nums[i] != last:
                path.append(nums[i])
                in_path[i] = True
                backtrack()
                last = path.pop()
                in_path[i] = False
        
        return
    backtrack()
        
    return result

nums = [1,1,2]
permuteUnique(nums)
````

### 39. 组合总和
>找出 candidates 中可以使数字和为目标数 target 的所有不同组合。

思路：加记录解的条件

和子集的不同点：元素可以无限制重复被选取

````python
def combinationSum(candidates, target):
    N = len(candidates)
    result = []
    def backtrack(start=0, path=[], path_sum = 0):
        if path_sum == target:
            result.append(path.copy())
            return
        if path_sum > target:
            return
        for i in range(start, N):
            # 尝试
            path.append(candidates[i])
            path_sum += candidates[i]
            # 进行下一轮选择
            backtrack(i,path,path_sum)
            # 回退
            path_sum -= path.pop()

    backtrack()
    return result
candidates = [2,3,6,7]
target = 7
combinationSum(candidates, target)
````

### 17. 电话号码的字母组合
>给定一个仅包含数字 2-9 的字符串，返回所有它能表示的字母组合。

思路：

````python
def letterCombinations(digits):
    if not digits: return []
    
    N = len(digits)
    result = []
    num2char = {
            '2': ['a', 'b', 'c'],
            '3': ['d', 'e', 'f'],
            '4': ['g', 'h', 'i'],
            '5': ['j', 'k', 'l'],
            '6': ['m', 'n', 'o'],
            '7': ['p', 'q', 'r', 's'],
            '8': ['t', 'u', 'v'],
            '9': ['w', 'x', 'y', 'z']
        }
    def backtrack(start=0,path=[]):
        if start == N:
            result.append(''.join(path))
            return
        for c in num2char[digits[start]]:
            path.append(c)
            backtrack(start+1)
            path.pop()    
    backtrack()
    
    return result
     
digits = "23"
letterCombinations(digits)
````

````python
import collections

def partition(s):
    palindrome = collections.defaultdict(set)
    def isPalindrome(i,j):
        if i<j:
            return j in palindrome[i]
        return True
    
    N = len(s)
    for j in range(N):
        for i in range(j+1):
            if s[i] == s[j] and isPalindrome(i+1, j-1):
                palindrome[i].add(j)
    print(palindrome)
    result = []
    def backtrack(start=0,path=[]):
        if start == N:
            result.append(path.copy())
            return
        
        for i in palindrome[start]:
            path.append(s[start:i+1])
            backtrack(i + 1)
            path.pop()

    backtrack()
    return result
    
s = "aab"
partition(s)
````
