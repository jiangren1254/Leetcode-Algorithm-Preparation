### 344. 反转字符串   
> 编写一个函数，其作用是将输入的字符串反转过来。输入字符串以字符数组char[]的形式给出。

- 思路1：双指针反转，最优。遍历一半的数组即可，原地修改。

递归分为两阶段  
- 递
- 归

```python
def reverseString(s):
    l,r = 0,len(s)-1
    while l<=r:
        s[l],s[r] = s[r],s[l]
        l += 1
        r -= 1
    
s = ["h","e","l","l","o"]
reverseString(s)
s
```

- 思路2：栈，次优。代码省略
- 思路3： s[::-1],简洁，但切片会新建一个列表，不满足在原地修改。代码省略
- 思路4：递归，次优。下面是递归的实现。实际上和思路是一样的。recursion(s,i,j)

```python
def reverseString(s):
    def recursion(s,i,j):
        if i >= j:
            return
        s[i],s[j] = s[j],s[i]
        recursion(s,i+1,j-1)
        
    recursion(s,0,len(s)-1)
s = ["h","e","l","l","o"]
reverseString(s)
s   
```

### 24. 两两交换链表中的节点
> 给定一个链表，两两交换其中相邻的节点，并返回交换后的链表。 你不能只是单纯的改变节点内部的值，而是需要实际的进行节点交换。

- 思路1：迭代实现。两两一对，其他的都是重复，处理完返回交换后的头结点

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
def swapPairs(head):
    # 正向遍历
    if not head or not head.next:
        return head
    tail = dummy = ListNode(next=head)
    prev, curr = head,head.next
    while prev and curr:
        prev.next = curr.next
        curr.next = prev
        tail.next = curr
        # post 处理
        tail = prev
        prev = tail.next
        if prev:
            curr = prev.next
    return dummy.next
      
```

- 递归实现。首先交换的是最后一对，这样挺巧的。递归实现简洁

```python
def swapPairs(head):
    if head and head.next:
        head_next_pair = swapPairs(head.next.next)
        nex = head.next
        head.next = head_next_pair
        nex.next = head
        head = nex
        
    return head
```

### 95. 不同的二叉搜索树 II
> 给定一个整数 n，生成所有由 1 ... n 为节点所组成的不同二叉搜索树。可以按任何顺序返回。

思路：n个节点的二叉搜索树维护，不好维护，但是2个节点，1个节点的好维护，因此递归每次处理的n-1

注意：此题用来训练递归思维有理论意义，但是实际上算法返回的树并不是 deep copy，多个树之间会共享子树。

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
def generateTrees(n):
    def generateTrees_recursion(i, j):
        if i > j: return[None]
        result = []
        for m in range(i,j+1):
            # 看不懂
            left = generateTrees_recursion(i,m-1)
            right = generateTrees_recursion(m+1,j)
            
            for l in left:
                for r in right:
                    result.append(TreeNode(m,l,r))
        return result
    
    return generateTrees_recursion(1,n) if n > 0 else []
    
    
    
```

### 509. 斐波那契数
> 斐波那契数，通常用  F(n) 表示，形成的序列称为斐波那契数列。  
> 该数列由  0 和 1 开始，后面的每一项数字都是前面两项数字的和。也就是： F(0) = 0,   F(1) = 1 F(N) = F(N - 1) + F(N - 2), 其中 N > 1. 给定  N，计算  F(N)。

- 下面是递归+记忆化实现的。

```python
def fib(n):
    mem = [-1] * (n + 2)   
    mem[0], mem[1] = 0, 1
    def fib_recursion(n):
        if mem[n] == -1:
            mem[n] = fib_recursion(n-1) + fib_recursion(n-2)
        return mem[n]
    
    return fib_recursion(n)

fib(4)
```

- 动态规划实现

```python
def fib(n: int) -> int:
    if n==0 or n==1: return n
    dp = [0,1]
    for i in range(2,n+1):
        dp[0],dp[1] = dp[1],dp[0]+dp[1]
    print(dp)
    return dp[1]
fib(4)
```

### 76. 最小覆盖子串
> 给你一个字符串 s、一个字符串 t，请在字符串 s 里面找出：包含 t 所有字母的最小子串

思路：没有顺序，
hard

```python
import collections
def minWindow( s: str, t: str) -> str:
    if len(t) > len(s): return ""

    target = collections.defaultdict(int)
    window = collections.defaultdict(int)
    for c in t:
        target[c] += 1
    # print(target)
    min_size = len(s) + 1
    min_str = ''
    left, right, count = 0,0, 0
    num_char = len(target)
    while right < len(s):
        c = s[right]  #c是将移入窗口的字符
        right += 1
        
        if c in target:
            # 进行窗口内一系列更新
            window[c] += 1
            if window[c] == target[c]:
                count += 1
                if count == num_char:
                    # print(f"window: [{left}, {right}]")
                    # print(window)
                    # 判断左侧窗口是否要收缩
                    while left <right and count == num_char:
                        # d 是将移除窗户的字符
                        d = s[left]
                        left += 1
                        if d in target:
                            # 进行窗口内一系列更新
                            window[d] -= 1
                            if window[d] == target[d] - 1:
                                count -= 1
                    # post
                    if min_size > right - left + 1:
                        min_size = right - left + 1
                        min_str = s[left - 1:right]    
        
    return min_str
s = "ADOBECODEBANC"
t = "ABC"
minWindow(s,t)
```

### 567. 字符串的排列
> 给定两个字符串 s1 和 s2，写一个函数来判断 s2 是否包含s1的排列。

- 思路：

permutation: 排列

```python
import collections
def checkInclusion(s1: str, s2: str) -> bool:
    target = collections.defaultdict(int)
    for c in s1:
        target[c] += 1
    r, num_char = 0,len(target)
    while r < len(s2):
        
        if s2[r] in target:
            l,count = r,0
            window = collections.defaultdict(int)
            while r < len(s2):
                c = s2[r]
                if c not in target:
                    break
                window[c] += 1
                if window[c] == target[c]:
                    count += 1
                    if count == num_char:
                        return True
                print(f"window: [{l}, {r}]")
                while window[c] > target[c]:
                    d = s2[l]
                    window[d] -= 1
                    if window[d] == target[d]-1:
                        count -= 1
                    l += 1
                # post
                r += 1
        else:
            r += 1
    return False
        
s1 = "ab"
s2 = "eidbaooo"
checkInclusion(s1, s2) 
```

### 438. 找到字符串中所有字母异位词
> 给定两个字符串 s 和 p，找到 s 中所有 p 的 异位词 的子串，返回这些子串的起始索引。不考虑答案输出的顺序。

思路：？

```python
def findAnagrams(s: str, p: str):
    target = collections.defaultdict(int)
        
    for c in p: target[c] += 1 
    
    r, num_char = 0, len(target)
    
    results = []
    while r < len(s):
        if s[r] in target:
            l, count = r, 0
            window = collections.defaultdict(int)
            while r < len(s):
                c = s[r]
                if c not in target:
                    break
                window[c] += 1
                if window[c] == target[c]:
                    count += 1
                    if count == num_char:
                        results.append(l)
                        window[s[l]] -= 1
                        count -= 1
                        l += 1
                while window[c] > target[c]:
                    window[s[l]] -= 1
                    if window[s[l]] == target[s[l]] - 1:
                        count -= 1
                    l += 1
                r += 1
        else:
            r += 1
    
    return results

s = "cbaebabacd"
p = "abc"
findAnagrams(s,p)
```

### 3. 无重复字符的最长子串
>给定一个字符串，请你找出其中不含有重复字符的   最长子串   的长度。 示例  1:  
输入: "abcabcbb" 输出: 3 解释: 因为无重复字符的最长子串是 "abc"，所以其长度为 3。

思路：遇到哈希表中的元素，收缩左侧窗口.下面这种实现比较高效

>note: 这题在力扣hot100 中，是最高频的


```python
def lengthOfLongestSubstring(s):
    last_idx = {}  # 记录字符出现的下标,后面的会把前面的覆盖掉
    l, ans = 0, 0
    for r, c in enumerate(s):
        if c in last_idx and last_idx[c] >=l:  #窗口收缩
            ans = max(ans, r-l)
            l = last_idx[c] + 1
        last_idx[c] = r      # 哈希记录

    # print(last_idx)
    return max(ans, len(s) - l) # 注意：最后一个子串不会在循环中被判断。
            
s = "abcabcbb"
lengthOfLongestSubstring(s)
```
