- s[::-1],简洁，但切片会新建一个列表

```python
def reverseString(s):
    l,r = 0,len(s)-1
    while l<r:
        s[l],s[r] = s[r],s[l]
        l += 1
        r -= 1
    
s = ["h","e","l","l","o"]
reverseString(s)
s
```

```python
def reverseString(s):
    def recursion(s,i,j):
        if i >= j:
            return
        s[i],s[j] = s[j],s[i]
        recursion(s,i+1,j-1)
    recursion(s,0,len(s)-1)
s = ["h","e","l","l","o"]
reverseString(s)
s   
```
